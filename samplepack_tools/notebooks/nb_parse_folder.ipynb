```python
from glob import glob
import os
```

```python
path = "/Users/carl/Downloads/mob"
output_path = "/Users/carl/dev/tsvr-samplepack-tools/data/all_mobs"

allowed_categories = [
    "bat",
    "chicken",
    "cow",
    "creeper",
    "horse",
    "pig",
    "pillager",
    "sheep",
    "skeleton",
    "spider",
    "villager",
    "zombie"
    ]
```

```python
files = glob(f"{path}/*/**.wav")
```

```python
all_files = []

for f in files:
    wav_name = f.split("/")[-1]
    category = f.split("/")[-2]
    if category not in allowed_categories:
        continue
    all_files.append(f"{wav_name}\t{category}")

    # copy the file to the new location
    os.system(f"cp {f} {output_path}/{category}_{wav_name.split('.')[0]}.wav")
```

```python
all_files = sorted(all_files)
```

```python
for f in all_files:
    print(f)
```
